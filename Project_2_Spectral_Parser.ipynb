```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter, find_peaks

# 1. Simulate a noisy UV-Vis Spectrophotometer signal
wavelengths = np.linspace(200, 700, 500)
true_signal = 1.2 * np.exp(-((wavelengths - 350)/30)**2) + 0.8 * np.exp(-((wavelengths - 500)/20)**2)
noisy_signal = true_signal + np.random.normal(0, 0.05, size=500)

# 2. Smooth the data using Savitzky-Golay Filter
smoothed_signal = savgol_filter(noisy_signal, window_length=21, polyorder=3)

# 3. Automatically detect absorption peaks
peaks, _ = find_peaks(smoothed_signal, height=0.3, distance=30)

# 4. Plot results
plt.figure(figsize=(10, 5))
plt.plot(wavelengths, noisy_signal, label='Raw Noisy Data', alpha=0.5, color='gray')
plt.plot(wavelengths, smoothed_signal, label='Cleaned/Smoothed Signal', color='blue', linewidth=2)
plt.plot(wavelengths[peaks], smoothed_signal[peaks], "x", label='Detected Peaks', color='red', markersize=10)
plt.xlabel('Wavelength (nm)')
plt.ylabel('Absorbance')
plt.title('Automated Spectral Data Parsing Workflow')
plt.legend()
plt.grid(True)
plt.show()

print("\n--- PROJECT 2 SUCCESSFUL ---")
print("Detected Peak Wavelengths (nm):", wavelengths[peaks].round(1))
```
