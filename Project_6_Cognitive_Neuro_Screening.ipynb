```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Simulate a Molecular Docking Screening Library for Cognitive Drug Targets
np.random.seed(101)
compounds = [f"Drug_Compound_{i:03d}" for i in range(1, 51)]

# Gibbs Free Energy Binding affinity (kcal/mol) - lower (more negative) means stronger cognitive binding
binding_energy = np.random.uniform(-9.5, -4.0, 50)
blood_brain_barrier_penetration = np.random.uniform(0.1, 0.95, 50) # BBB Score

df = pd.DataFrame({
    'Compound_ID': compounds,
    'Binding_Affinity_kcal_mol': binding_energy,
    'BBB_Penetration_Score': blood_brain_barrier_penetration
})

# 2. Apply Clinical Research Screening Criteria
# Strong hit = Binding Energy <= -7.5 kcal/mol AND BBB score >= 0.50
def screen_cognitive_leads(row):
    if row['Binding_Affinity_kcal_mol'] <= -7.5 and row['BBB_Penetration_Score'] >= 0.50:
        return 'High-Affinity Cognitive Lead'
    return 'Candidate Filtered Out'

df['Screening_Result'] = df.apply(screen_cognitive_leads, axis=1)

# 3. Print Lead Discoveries
leads = df[df['Screening_Result'] == 'High-Affinity Cognitive Lead']
print("--- HEALTHCARE & COGNITIVE PROJECT SUCCESSFUL ---")
print(f"Total Compounds Screened: {len(df)}")
print(f"Potential Cognitive Therapeutics Discovered: {len(leads)}")
print("\nTop 5 Discovered Drug Candidates for Neuro-Targeting:")
print(leads.sort_values(by='Binding_Affinity_kcal_mol').head().to_string(index=False))

# 4. Generate the Healthcare Analytics Heatmap
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Binding_Affinity_kcal_mol', y='BBB_Penetration_Score', hue='Screening_Result', palette={'High-Affinity Cognitive Lead': 'magenta', 'Candidate Filtered Out': 'gray'}, s=80)
plt.axvline(x=-7.5, color='black', linestyle='--', alpha=0.5)
plt.axhline(y=0.5, color='black', linestyle='--', alpha=0.5)
plt.title('In-Silico Screening Matrix: Neurological Drug Candidates vs. Cognitive Receptors')
plt.xlabel('Binding Affinity (Gibbs Free Energy ΔG in kcal/mol)')
plt.ylabel('Blood-Brain Barrier (BBB) Penetration Propensity')
plt.grid(True, alpha=0.2)
plt.show()
```
