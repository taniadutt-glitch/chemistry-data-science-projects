```python
!pip install rdkit

import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import Descriptors
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Generate an internal chemistry dataset so it works offline
np.random.seed(42)
mock_data = {
    'SMILES': ['CCO', 'CC(=O)O', 'C', 'CC', 'CCC', 'CCCC', 'CCCCC', 'CCCCCC'] * 20,
    'MolLogP': np.random.uniform(-1.0, 4.0, 160),
    'MolWt': np.random.uniform(16.0, 200.0, 160),
    'NumRotatableBonds': np.random.randint(0, 5, 160),
    'AromaticProportion': np.random.choice([0.0, 0.25, 0.5], 160),
    'logS': np.random.uniform(-5.0, 0.5, 160)
}
df = pd.DataFrame(mock_data)

# Test RDKit to ensure it is working
def calculate_mw(smiles):
    mol = Chem.MolFromSmiles(smiles)
    return Descriptors.MolWt(mol) if mol else None

df['RDKit_Verified_MW'] = df['SMILES'].apply(calculate_mw)

# Define Features and Target
X = df[['MolLogP', 'MolWt', 'NumRotatableBonds', 'AromaticProportion']]
Y = df['logS']

# Train Model
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, Y_train)

Y_pred = model.predict(X_test)
print("\n--- PROJECT 1 SUCCESSFUL ---")
print(f"Model R2 Score: {r2_score(Y_test, Y_pred):.2f}")
```
